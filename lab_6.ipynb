

## 1. Множества и параметры

У нас есть ширина листа:

$$
W=13.
$$

Есть высота исходного листа:

$$
H_0=19.
$$

Можно задать множество типов прямоугольников:

$$
I=\{1,2,3,4,5\}.
$$

Для каждого типа $i\in I$ заданы параметры:

- $a_i$ — ширина прямоугольника типа $i$;
- $b_i$ — высота прямоугольника типа $i$;
- $c_i$ — доход от продажи одного прямоугольника типа $i$.

Значения параметров:

| $i$ | $a_i$ | $b_i$ | $c_i$ |
|---|---|---|---|
| 1 | 4 | 3 | 20 |
| 2 | 3 | 5 | 18 |
| 3 | 5 | 7 | 30 |
| 4 | 7 | 2 | 20 |
| 5 | 3 | 3 | 12 |

Так как на первом этапе материал разрезается параллельно оси $y$, получаются вертикальные полосы.

Поскольку повороты прямоугольников запрещены, множество возможных ширин полос:

$$
K=\{3,4,5,7\}.
$$

## 2. Множество шаблонов раскроя

Рассмотрим произвольную вертикальную полосу.

Введём множество $P(H)$ — множество всех допустимых шаблонов раскроя одной полосы при заданной высоте листа $H$.

**Шаблон раскроя** — это конкретная комбинация прямоугольников, расположенных друг над другом в одной вертикальной полосе.

Каждый шаблон $p\in P(H)$ характеризуется следующими параметрами:

- $w_p$ — ширина полосы;
- $n_{ip}$ — количество прямоугольников типа $i$ в шаблоне $p$;
- $h_p$ — суммарная высота прямоугольников в полосе;
- $r_p$ — суммарный доход от всех прямоугольников в полосе;
- $s_p$ — суммарная площадь прямоугольников в полосе.

Параметры шаблона вычисляются следующим образом.

Занятая высота полосы:

$$
h_p=\sum_{i\in I}b_i n_{ip}.
$$

Суммарный доход от изделий в полосе:

$$
r_p=\sum_{i\in I}c_i n_{ip}.
$$

Суммарная площадь изделий в полосе:

$$
s_p=\sum_{i\in I}a_i b_i n_{ip}.
$$

### Условия допустимости шаблона

Шаблон $p$ является допустимым, если выполняются следующие условия.

**1. Ограничение по высоте полосы:**

$$
\sum_{i\in I}b_i n_{ip}\le H.
$$

**2. Совпадение ширины прямоугольников с шириной полосы:**

$$
n_{ip}=0,\quad \forall i\in I: a_i\ne w_p.
$$

То есть в одной полосе могут находиться только прямоугольники, ширина которых совпадает с шириной полосы.

**3. Неотрицательность и целочисленность количества изделий:**

$$
n_{ip}\in\mathbb Z_{\ge0},
\quad \forall i\in I.
$$

**4. Шаблон должен содержать хотя бы один прямоугольник:**

$$
\sum_{i\in I}n_{ip}\ge1.
$$

Таким образом, множество $P(H)$ формируется из всех комбинаций прямоугольников, удовлетворяющих перечисленным условиям.

### Пример шаблона

Пусть рассматривается полоса шириной $w_p=3$, содержащая:

- 2 прямоугольника типа 2 размером $3\times5$;
- 3 прямоугольника типа 5 размером $3\times3$.

Тогда:

$$
(n_{1p},n_{2p},n_{3p},n_{4p},n_{5p})
=(0,2,0,0,3).
$$

Занятая высота:

$$
h_p=2\cdot5+3\cdot3=19.
$$

Доход:

$$
r_p=2\cdot18+3\cdot12=72.
$$

Площадь:

$$
s_p=2\cdot15+3\cdot9=57.
$$

Следовательно, данный шаблон принадлежит множеству $P(19)$.

## 3. Переменные решения

Введём целочисленную переменную:

$$
x_p\in\mathbb Z_{\ge0},\quad p\in P(H_0),
$$

где $x_p$ — количество вертикальных полос, раскроенных по шаблону $p$.

Например, если $x_p=3$, то рассматриваемый шаблон используется три раза.

Дополнительно определим количество изготовленных прямоугольников каждого типа:

$$
z_i=\sum_{p\in P(H_0)}n_{ip}x_p,
\quad i\in I.
$$

Здесь $z_i$ — суммарное количество изготовленных прямоугольников типа $i$ по всем полосам.

Переменные $z_i$ являются вспомогательными и могут быть выражены непосредственно через $x_p$.

## 4. Ограничения модели

### 4.1. Ограничение по ширине листа

Суммарная ширина всех используемых полос не должна превышать ширину исходного листа:

$$
\boxed{
\sum_{p\in P(H_0)}w_p x_p\le W.
}
$$

(Продолжим наш пример):

$$
\sum_{p\in P(19)}w_p x_p\le13.
$$

### 4.2. Ограничение количества полос каждого шаблона

Максимальное количество полос шириной $w_p$, которое можно получить из листа шириной $W$:

$$
U_p=\left\lfloor\frac{W}{w_p}\right\rfloor.
$$

Следовательно:

$$
\boxed{
0\le x_p\le U_p,
\quad x_p\in\mathbb Z.
}
$$

### 4.3. Ограничение по высоте

Для каждого шаблона должно выполняться:

$$
h_p\le H_0.
$$

Но так как мы это учитываем при создании множества $P(H)$, то отдельно прописывать его не будем.

## 5. ЗМаксимизация суммарного дохода

Необходимо найти такой план раскроя листа $13\times19$, при котором суммарный доход от продажи прямоугольников будет максимальным.

Общий доход:

$$
R=\sum_{i\in I}c_i z_i.
$$

Подставляя выражение для $z_i$, получаем:

$$
R=\sum_{p\in P(H_0)}r_p x_p.
$$

### Целевая функция

$$
\boxed{
R\rightarrow\max.
}
$$

### Тогда:

$$
\boxed{
\begin{aligned}
\max\quad&
\sum_{p\in P(H_0)}r_p x_p\\[2mm]
\text{при условиях:}\quad&
\sum_{p\in P(H_0)}w_p x_p\le W,\\[2mm]
&0\le x_p\le U_p,\quad p\in P(H_0),\\
&x_p\in\mathbb Z,\quad p\in P(H_0).
\end{aligned}
}
$$

## 6. Минимизация отходов

Необходимо найти план раскроя, при котором площадь неиспользованного материала будет минимальной.

Площадь исходного листа:

$$
S_0=WH_0=13\cdot19=247.
$$

Суммарная площадь изготовленных прямоугольников:

$$
S=\sum_{p\in P(H_0)}s_p x_p.
$$

Площадь отходов:

$$
D=S_0-S.
$$

Следовательно:

$$
D=WH_0-\sum_{p\in P(H_0)}s_p x_p.
$$

### Целевая функция

$$
\boxed{
D\rightarrow\min.
}
$$

Поскольку площадь исходного листа постоянна, минимизация отходов эквивалентна максимизации площади изготовленных прямоугольников:

$$
\boxed{
\sum_{p\in P(H_0)}s_p x_p\rightarrow\max.
}
$$

### Тогда

$$
\boxed{
\begin{aligned}
\max\quad&
\sum_{p\in P(H_0)}s_p x_p\\[2mm]
\text{при условиях:}\quad&
\sum_{p\in P(H_0)}w_p x_p\le W,\\[2mm]
&0\le x_p\le U_p,\quad p\in P(H_0),\\
&x_p\in\mathbb Z,\quad p\in P(H_0).
\end{aligned}
}
$$

## 7. Минимизация высоты листа

Теперь необходимо определить минимальную высоту листа при выполнении следующих условий:

- ширина листа остаётся равной $W=13$;
- прямоугольников каждого типа можно изготовить не более 5 штук;
- суммарный доход должен составлять не менее 200 рублей.

### 7.1. Дополнительные переменные

Введём целочисленную переменную:

$$
H\in\mathbb Z,
$$

где $H$ — искомая высота листа.

Поскольку при высоте 19 допустимый план существует, можем ограничить её сверху:

$$
1\le H\le19.
$$

Для третьей задачи используем множество заранее сгенерированных шаблонов $P(H_0)=P(19)$.

Введём бинарную переменную:

$$
y_p\in\{0,1\},
\quad p\in P(H_0),
$$

где

$$
y_p=
\begin{cases}
1,&\text{если шаблон }p\text{ используется};\\
0,&\text{если шаблон }p\text{ не используется}.
\end{cases}
$$

### 7.2. Дополнительные ограничения

**1. Ограничение по количеству изделий каждого типа:**

$$
\boxed{
\sum_{p\in P(H_0)}n_{ip}x_p\le5,
\quad\forall i\in I.
}
$$

**2. Ограничение минимального дохода:**

$$
\boxed{
\sum_{p\in P(H_0)}r_p x_p\ge200.
}
$$

**3. Связь количества полос с фактом использования шаблона:**

$$
\boxed{
y_p\le x_p\le U_p y_p,
\quad\forall p\in P(H_0).
}
$$

Если шаблон не используется, $y_p=0$ и $x_p=0$.

Если шаблон используется, $y_p=1$ и $x_p\ge1$.

**4. Ограничение высоты используемых шаблонов:**

$$
\boxed{
H\ge h_p y_p,
\quad\forall p\in P(H_0).
}
$$

Если шаблон $p$ используется, его высота не может превышать высоту искомого листа.

### 7.3. Целевая функция

$$
\boxed{
H\rightarrow\min.
}
$$

### Тогда

$$
\boxed{
\begin{aligned}
\min\quad &H\\[2mm]
\text{при условиях:}\quad
&\sum_{p\in P(H_0)}w_p x_p\le W,\\[2mm]
&\sum_{p\in P(H_0)}n_{ip}x_p\le5,
&&\forall i\in I,\\[2mm]
&\sum_{p\in P(H_0)}r_p x_p\ge200,\\[2mm]
&y_p\le x_p,
&&\forall p\in P(H_0),\\
&x_p\le U_p y_p,
&&\forall p\in P(H_0),\\
&H\ge h_p y_p,
&&\forall p\in P(H_0),\\[2mm]
&1\le H\le H_0,\\
&H\in\mathbb Z,\\
&x_p\in\mathbb Z_{\ge0},
&&\forall p\in P(H_0),\\
&y_p\in\{0,1\},
&&\forall p\in P(H_0).
\end{aligned}
}
$$

## 8. Пример негильотинного раскроя

Негильотинным называется раскрой, при котором невозможно разделить материал на требуемые прямоугольники последовательностью сквозных прямолинейных разрезов от одной границы текущего фрагмента до противоположной.

Рассмотрим четыре прямоугольника, расположенные внутри области $10\times7$ исходного листа.

Координаты левых нижних углов и размеры прямоугольников:

| Прямоугольник | Тип | $x$ | $y$ | Размер |
|---|---|---|---|---|
| 1 | 4 | 0 | 0 | $7\times2$ |
| 2 | 2 | 7 | 0 | $3\times5$ |
| 3 | 4 | 3 | 5 | $7\times2$ |
| 4 | 2 | 0 | 2 | $3\times5$ |

Все четыре прямоугольника не пересекаются и помещаются внутри исходного листа $13\times19$.

Однако внутри области $10\times7$ нельзя провести сквозной вертикальный или горизонтальный разрез, который не пересекал бы ни одного из четырёх прямоугольников.

Следовательно, данная конфигурация является примером негильотинного раскроя.

## 9. Метод решения

Для решения задач 1–3 используется метод шаблонов раскроя.

Алгоритм состоит из следующих этапов:

1. Формирование множества допустимых ширин полос $K$.
2. Генерация всех допустимых шаблонов $P(H_0)$.
3. Вычисление параметров $w_p,n_{ip},h_p,r_p,s_p$ для каждого шаблона.
4. Создание целочисленных переменных $x_p$.
5. Построение ограничений математической модели.
6. Формирование целевой функции в зависимости от задачи.
7. Решение полученной MILP-модели с помощью SCIP.
8. Восстановление плана раскроя по найденным значениям $x_p$.

В третьей задаче дополнительно вводятся переменная высоты $H$ и бинарные переменные $y_p$.

```python

from dataclasses import dataclass
from itertools import product
from ortools.linear_solver import pywraplp


# исходные данные
W, H0 = 13, 19

# (a_i, b_i, c_i): ширина, высота, цена изделия типа i
ITEMS = [
    (4, 3, 20),
    (3, 5, 18),
    (5, 7, 30),
    (7, 2, 20),
    (3, 3, 12),
]

I = range(len(ITEMS))
K = sorted({a for a, b, c in ITEMS})


@dataclass(frozen=True)
class Pattern:
    width: int               # w_p — ширина полосы
    counts: tuple[int, ...]  # n_ip — количество изделий каждого типа
    height: int              # h_p — занятая высота полосы
    revenue: int             # r_p — доход от одной полосы
    area: int                # s_p — полезная площадь одной полосы


def generate_patterns(H):
    """Формирует P(H): все непустые допустимые шаблоны одной полосы."""
    patterns = []

    for w in K:
        ids = [i for i in I if ITEMS[i][0] == w]
        ranges = [range(H // ITEMS[i][1] + 1) for i in ids]

        for numbers in product(*ranges):
            counts = [0] * len(ITEMS)

            for i, n in zip(ids, numbers):
                counts[i] = n

            if sum(counts) == 0:
                continue

            hp = sum(ITEMS[i][1] * counts[i] for i in I)

            if hp > H:
                continue

            rp = sum(ITEMS[i][2] * counts[i] for i in I)

            sp = sum(
                ITEMS[i][0] * ITEMS[i][1] * counts[i]
                for i in I
            )

            patterns.append(
                Pattern(w, tuple(counts), hp, rp, sp)
            )

    return patterns


def base_model(P):
    """Общие переменные x_p и ограничение ширины листа."""
    solver = pywraplp.Solver.CreateSolver("SCIP")

    # x_p — количество полос шаблона p
    # U_p = floor(W / w_p)
    x = [
        solver.IntVar(0, W // p.width, f"x_{j}")
        for j, p in enumerate(P)
    ]

    # ограничение ширины:
    # sum(w_p * x_p) <= W
    solver.Add(
        solver.Sum(
            p.width * x[j]
            for j, p in enumerate(P)
        ) <= W
    )

    return solver, x


def solve_and_print(solver, P, x, height):
    status = solver.Solve()

    status_names = {
        pywraplp.Solver.OPTIMAL: "OPTIMAL",
        pywraplp.Solver.FEASIBLE: "FEASIBLE (оптимальность не доказана)",
        pywraplp.Solver.INFEASIBLE: "INFEASIBLE",
        pywraplp.Solver.UNBOUNDED: "UNBOUNDED",
        pywraplp.Solver.ABNORMAL: "ABNORMAL",
        pywraplp.Solver.NOT_SOLVED: "NOT_SOLVED",
    }

    if status != pywraplp.Solver.OPTIMAL:
        raise RuntimeError(
            f"SCIP не подтвердил оптимальное решение: "
            f"{status_names.get(status, status)}"
        )

    H = (
        height if isinstance(height, int)
        else round(height.solution_value())
    )

    selected = [
        (p, round(x[j].solution_value()))
        for j, p in enumerate(P)
        if x[j].solution_value() > 0.5
    ]

    counts = [
        sum(p.counts[i] * n for p, n in selected)
        for i in I
    ]

    revenue = sum(
        p.revenue * n for p, n in selected
    )

    area = sum(
        p.area * n for p, n in selected
    )

    used_width = sum(
        p.width * n for p, n in selected
    )

    waste = W * H - area

    # Геометрическая проверка решения
    assert used_width <= W
    assert all(p.height <= H for p, n in selected)
    assert waste >= 0

    print(f"Статус: {status_names[status]}")
    print(f"Размер листа: {W} x {H}")
    print(f"Использовано ширины: {used_width} из {W}")
    print(f"Доход: {revenue} руб.")
    print(f"Отходы: {waste} ед.^2")
    print(f"Количество изделий по типам 1..5: {counts}")

    print("Выбранные шаблоны:")

    for p, n in selected:
        print(
            f"  x_p={n}; w_p={p.width}; n_ip={p.counts}; "
            f"h_p={p.height}; r_p={p.revenue}; s_p={p.area}"
        )

    return {
        "height": H,
        "revenue": revenue,
        "waste": waste,
        "counts": counts,
        "patterns": selected,
    }


def task1(P):
    """Пункт 1: максимизировать суммарный доход."""
    solver, x = base_model(P)

    # max sum(r_p * x_p)
    solver.Maximize(
        solver.Sum(
            p.revenue * x[j]
            for j, p in enumerate(P)
        )
    )

    print("\n1. МАКСИМИЗАЦИЯ ДОХОДА")

    return solve_and_print(solver, P, x, H0)


def task2(P):
    """Пункт 2: минимизировать площадь отходов."""
    solver, x = base_model(P)

    # min отходы <=> max полезная площадь
    # max sum(s_p * x_p)
    solver.Maximize(
        solver.Sum(
            p.area * x[j]
            for j, p in enumerate(P)
        )
    )

    print("\n2. МИНИМИЗАЦИЯ ОТХОДОВ")

    return solve_and_print(solver, P, x, H0)


def task3(P):
    """Пункт 3: min H при z_i <= 5 и доходе >= 200 руб."""
    solver, x = base_model(P)

    # искомая высота листа
    H = solver.IntVar(1, H0, "H")

    # y_p — факт использования шаблона p
    y = [
        solver.BoolVar(f"y_{j}")
        for j in range(len(P))
    ]

    for j, p in enumerate(P):
        Up = W // p.width

        # y_p <= x_p <= U_p * y_p
        solver.Add(y[j] <= x[j])
        solver.Add(x[j] <= Up * y[j])

        # H >= h_p * y_p
        solver.Add(H >= p.height * y[j])

    # не более 5 изделий каждого типа:
    # z_i = sum(n_ip * x_p) <= 5
    for i in I:
        solver.Add(
            solver.Sum(
                p.counts[i] * x[j]
                for j, p in enumerate(P)
            ) <= 5
        )

    # суммарный доход не менее 200 рублей:
    # sum(r_p * x_p) >= 200
    solver.Add(
        solver.Sum(
            p.revenue * x[j]
            for j, p in enumerate(P)
        ) >= 200
    )

    # минимизация высоты
    solver.Minimize(H)

    print("\n3. МИНИМИЗАЦИЯ ВЫСОТЫ")

    result = solve_and_print(solver, P, x, H)

    # проверим дополнительные ограничения
    assert all(n <= 5 for n in result["counts"])
    assert result["revenue"] >= 200

    return result


def task4():
    """Пункт 4: геометрический пример негильотинного раскроя."""

    # (x, y, ширина, высота, тип изделия [1..5])
    rectangles = [
        (0, 0, 7, 2, 4),
        (7, 0, 3, 5, 2),
        (3, 5, 7, 2, 4),
        (0, 2, 3, 5, 2),
    ]

    # проверим соответствие исходным типам,
    # границам листа и отсутствие пересечений
    for x, y, w, h, typ in rectangles:
        assert (w, h) == ITEMS[typ - 1][:2]
        assert x >= 0 and y >= 0
        assert x + w <= W and y + h <= H0

    for j, (x1, y1, w1, h1, _) in enumerate(rectangles):
        for x2, y2, w2, h2, _ in rectangles[j + 1:]:
            overlap = (
                max(x1, x2) < min(x1 + w1, x2 + w2)
                and max(y1, y2) < min(y1 + h1, y2 + h2)
            )

            assert not overlap, "Прямоугольники пересекаются"

    # рассмотрим область 10x7,
    # в которой расположены четыре изделия
    core_w, core_h = 10, 7

    xs = sorted(
        {0, core_w}
        | {x for x, y, w, h, typ in rectangles}
        | {x + w for x, y, w, h, typ in rectangles}
    )

    ys = sorted(
        {0, core_h}
        | {y for x, y, w, h, typ in rectangles}
        | {y + h for x, y, w, h, typ in rectangles}
    )

    # проверим потенциальные разрезы:
    # границы прямоугольников и промежутки между ними
    def candidate_cuts(edges):
        return sorted(
            set(edges[1:-1])
            | {
                (a + b) / 2
                for a, b in zip(edges, edges[1:])
            }
        )

    # существует ли вертикальный сквозной разрез?
    vertical_possible = any(
        all(
            not (x < t < x + w)
            for x, y, w, h, typ in rectangles
        )
        for t in candidate_cuts(xs)
        if 0 < t < core_w
    )

    # существует ли горизонтальный сквозной разрез?
    horizontal_possible = any(
        all(
            not (y < t < y + h)
            for x, y, w, h, typ in rectangles
        )
        for t in candidate_cuts(ys)
        if 0 < t < core_h
    )

    # если нет ни вертикального, ни горизонтального
    # первого разреза, конфигурация негильотинная
    assert not vertical_possible and not horizontal_possible, (
        "В центральном блоке существует сквозной разрез"
    )

    print("\n4. ПРИМЕР НЕГИЛЬОТИННОГО РАСКРОЯ")

    for x, y, w, h, typ in rectangles:
        print(
            f"  Тип {typ}: ({x}, {y}), размер {w} x {h}"
        )

    print(
        "В блоке 10 x 7 нет допустимого "
        "первого сквозного разреза."
    )



P = generate_patterns(H0)

print(f"Число допустимых шаблонов P({H0}): {len(P)}")

# Для исходных данных должно получиться 34 шаблона
assert len(P) == 34

task1(P)
task2(P)
task3(P)
task4()
```
