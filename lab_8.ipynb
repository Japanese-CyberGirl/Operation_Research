```python
from pathlib import Path
import re

current_dir = Path.cwd().resolve()

PATH = current_dir / "data" / "resourse.txt"
```

```python
def read_data(path: Path):
    text = path.read_text(encoding="utf-8")

    sections = [
        section.strip()
        for section in text.split("\n\n")
        if section.strip()
    ]

    K, J = map(int, re.findall(r"\d+", sections[0]))

    q_k = list(map(int, sections[1].splitlines()[1:]))

    r_jk = [
        list(map(int, line.split()))
        for line in sections[2].splitlines()[1:]
    ]

    ro_j = list(map(int, sections[3].splitlines()[1:]))

    d_j = list(map(int, sections[4].splitlines()[1:]))

    pred = [
        tuple(map(int, line.split("->")))
        for line in sections[5].splitlines()[1:]
    ]

    return K, J, q_k, r_jk, ro_j, d_j, pred


K, J, q_k, r_jk, ro_j, d_j, pred = read_data(PATH)
```

```python
print(f"K = {K}")
print(f"J = {J}")
print(f"q_k(t) = {q_k}")
print(f"r_jk(tau) = {r_jk}")
print(f"ro_j = {ro_j}")
print(f"d_j = {d_j}")
print(f"pred = {pred}")
```

$K=5$ - количество разных ресурсов  
$J=50$  - множество работ  
$q_k(t)$ - количество ресурса $k$, выделяемое в момент времени $t$ на выполнение **всех** работ  
$r_{j,k}(\tau)$ - потребность в ресурсе $k$ в интервале времени $\[\tau - 1, \tau \], считая от начала выполнения работы $j$.  
$\rho_j$ - длительность работы $j$  
$d_j$ - директивный срок окончания работы $j$  
pred - отношение предшествования
